# Variation and Standard Variation


```python
import numpy as np
a = np.array([48 , 50 , 52])
b = np.array([10 , 50 , 80])

amean = np.mean(a)
bmean = np.mean(b)
print("Mean of A is : ",amean)
print("Mean of A is : ",amean)
```

```python
arange  = np.max(a) - np.min(a)
print("Range of the Array A is : ",arange)

brange  = np.max(b) - np.min(b)
print("Range of the Array B is : ",brange)
```

```python
variation = np.var(a)
stdvariation = np.std(a)
print(f"variation of Array A is : {variation}\n Standard variation of the Array A : {stdvariation}")
variation = np.var(b)
stdvariation = np.std(b)
print(f"variation of Array B is : {variation}\n Standard variation of the Array B : {stdvariation}")
```
