```python
sample_text = {"../documents/txt_files/AI_intro.txt":"""
Artificial Intelligence in Healthcare

Artificial Intelligence (AI) has become one of the most transformative technologies in modern healthcare. By leveraging machine learning, natural language processing, and computer vision, AI systems can assist medical professionals in diagnosis, treatment planning, patient monitoring, and administrative workflows.

History of AI in Medicine

The application of AI in healthcare began in the 1970s with expert systems such as MYCIN, which was developed at Stanford University. MYCIN was designed to identify bacterial infections and recommend antibiotic treatments. Although it performed well in controlled settings, it was never widely adopted due to legal and ethical concerns.

During the 1990s and early 2000s, healthcare institutions focused primarily on digitizing patient records. This led to the adoption of Electronic Health Records (EHRs), creating large datasets that later enabled machine learning models to be trained more effectively.

Major Applications
Medical Imaging

AI-powered systems are widely used in radiology. Deep learning models can analyze X-rays, CT scans, and MRIs to detect abnormalities such as tumors, fractures, and lesions. In some cases, AI systems have demonstrated performance comparable to experienced radiologists.

Drug Discovery

Pharmaceutical companies use AI to accelerate drug development. Traditional drug discovery may take 10 to 15 years, while AI-based screening can reduce early-stage research time significantly by predicting molecular interactions.

Predictive Analytics

Hospitals use predictive models to identify high-risk patients. These models analyze variables such as age, blood pressure, heart rate, and medical history to estimate the likelihood of complications such as sepsis or heart failure.

Benefits

AI offers several benefits to healthcare systems:

Improved diagnostic accuracy
Faster processing of medical data
Reduced administrative burden
Enhanced patient outcomes
Lower long-term operational costs

AI also enables personalized medicine by tailoring treatments to individual patient characteristics, including genetic profiles.

Challenges

Despite its promise, AI in healthcare faces several challenges.

Data Privacy

Medical data is highly sensitive. Organizations must comply with regulations such as HIPAA in the United States and GDPR in Europe to protect patient privacy.

Bias

AI models may produce biased outcomes if training data lacks diversity. This can lead to disparities in diagnosis and treatment across different populations.

Explainability

Many deep learning models function as black boxes, making it difficult for clinicians to understand how a prediction was generated.

Future Outlook

The future of AI in healthcare is expected to include more autonomous systems, remote patient monitoring, wearable integrations, and real-time diagnostics. Experts predict that human-AI collaboration, rather than full automation, will define the next decade of healthcare innovation.

In conclusion, AI has the potential to revolutionize healthcare by improving efficiency, reducing costs, and supporting better patient care. However, responsible deployment remains critical for long-term success."""}



for filepath,content in sample_text.items():
    with open(filepath,'w',encoding="utf-8") as f:
        f.write(content)

print("Sample text file Successfully created !!!")
```
