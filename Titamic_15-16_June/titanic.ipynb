# Data Cleaning 


```python
import numpy as np
import pandas as pd 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

data = pd.read_csv('Titanic-Dataset.csv')
data.isnull().sum()
data.shape
```

```python
#filling the null value in the age
data['Age'] = data['Age'].ffill()
data.isnull().sum()
```

```python
# drop the unwanted columns 
data = data.drop(columns=['Cabin' , 'Embarked'],)
data.to_csv('Cleaned_titanic.csv')
data.shape
```

```python
data.shape
```

```python
duplicate = data.duplicated().sum()
print(duplicate)
```

```python
survived = (data['Survived'] == 1).sum()
notsurvived = (data['Survived'] == 0).sum()
total = data['Survived'].count()
print("\n survived : ", survived)
print("\n Not survived : ", notsurvived)
print("\n Total Pasenger  : ", total)

```

```python
#Survival percentage 
per = (survived*100)/total
nper = (notsurvived*100)/total
print("Survival ",round(per,2),"%")
print("Non Survival ",round(nper,2),"%")
```

```python
Survival_group_by = data.groupby('Sex')['Survived'].mean()*100
print(round(Survival_group_by,2))
```

```python
Survival_group_by_ticket_type = data.groupby('Pclass')['Survived'].mean()*100
print(round(Survival_group_by_ticket_type,2))
```
