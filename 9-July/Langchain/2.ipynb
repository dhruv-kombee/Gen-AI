# Embedding models 

```python
# Import Libraries
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings

load_dotenv()
```

```python
# Initialize Gemini Embedding Model
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)
```

```python
# Embed a Single Text
text = "Artificial Intelligence is transforming the world."

vector = embeddings.embed_query(text)

print(f"Vector Dimension : {len(vector)}")
print(f"First 10 Values  : {vector[:10]}")
```

```python
documents = [
    "Artificial Intelligence is transforming the world.",
    "Machine Learning is a subset of AI.",
    "Deep Learning uses neural networks.",
    "LangChain simplifies LLM application development."
]

document_vectors = embeddings.embed_documents(documents)

print(f"Number of Documents : {len(document_vectors)}")
print(f"Embedding Dimension : {len(document_vectors[0])}")

print("\nFirst 10 values of First Document:")
print(document_vectors[0][:10])
```

```python
for i, vector in enumerate(document_vectors):
    print(f"Document {i+1} -> Dimension: {len(vector)}")
```

```python
query = input("Enter a sentence: ")

query_embedding = embeddings.embed_query(query)

print("\nEmbedding Dimension:", len(query_embedding))
print("First 10 Values:", query_embedding[:10])
```
