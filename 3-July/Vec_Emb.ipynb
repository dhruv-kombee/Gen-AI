## Module 1: Vector Databases

A vector database stores embeddings (high-dimensional vectors)
for efficient similarity search.

Examples:
- FAISS
- Pinecone
- Weaviate

```python
documents = [
    "FAISS is a vector similarity search library.",
    "Pinecone is a managed vector database.",
    "Weaviate supports hybrid semantic search.",
    "RAG combines retrieval with generation."
]
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(documents)

print(embeddings.shape)
```

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings).astype("float32"))

print(index.ntotal)
```

```python
query = "semantic retrieval database"

query_embedding = model.encode([query]).astype("float32")

distances, indices = index.search(query_embedding, 2)

for i in indices[0]:
    print(documents[i])
```
