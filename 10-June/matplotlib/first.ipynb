# Fundamentals of the Matplotlib 

```python
import matplotlib.pyplot as plt 

x = [1 , 2 , 3 , 4 , 5]
y = [20 , 10 , 30 , 20 , 40]

plt.plot(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt 

x = ["Mon", "Tue" , "Wed" , "Thu" , "Fri"]
y = [20 , 10 , 30 , 20 , 40]


plt.plot(x,y)
plt.title("Sale of the product in the first 5 day in $10k")
plt.xlabel("First Five days")
plt.ylabel("Sales of the product")
plt.show()
```

```python
x = ["Mon", "Tue" , "Wed" , "Thu" , "Fri"]
y = [20 , 10 , 30 , 20 , 40]

plt.plot(x,y,label = '2025' , marker = 'o')
plt.title("Sale of the product in the first 5 day in $10k")
plt.xlabel("First Five days")
plt.ylabel("Sales of the product")
plt.legend(loc='upper left')
plt.grid()
plt.ylim(0 ,50)

plt.show()
```
