# Model Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score ,f1_score

y_true = [1,0,1,1,0,0,1]
y_predict = [1,0,1,0,0,1,1]

print("Accuray Score: ", accuracy_score(y_true,y_predict))
print("Precision Score : ", precision_score(y_true,y_predict))
print("Recall Score : ", recall_score(y_true,y_predict))
print("F1 Score : ", f1_score(y_true,y_predict))
```

```python
from sklearn.metrics import confusion_matrix
y_true = [1,0,0,1,0,1,0,1,1,0]
y_pred = [1,0,1,1,0,1,0,0,1,0]
cm = confusion_matrix(y_true,y_pred)
print("Confusion Matrix : \n")
print(cm)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error

import numpy as np 

real_score = [90,60,80,100]
predicted_scores = [85,70,70,95]

mae = mean_absolute_error(real_score,predicted_scores)
mse = mean_squared_error(real_score,predicted_scores)
rmse = np.sqrt(mse)

print("MAE : on average off by : ",mae)
print("MSE : on Sqaure mistake value : ",mse)
print("RMSE : Final Realistic error : ",rmse)
```
