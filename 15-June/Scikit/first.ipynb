```python
from sklearn.linear_model import LinearRegression

X = [[1],[2],[3],[4],[5]]
y = [40,50,60,70,80]

model = LinearRegression()
model.fit(X,y)
hours = float(input("Enter the hours you studied  : "))
predicted_marks = model.predict([[hours]])
print(f"If you Studied {hours} the you'll be score arround : {predicted_marks}")
```

```python
from sklearn.linear_model import LogisticRegression

X = [[1],[2],[3],[4],[5]]
y = [0,0,1,1,1]

model = LogisticRegression()
model.fit(X,y)
hours = float(input("Enter the hours you studied : "))
res = model.predict([[hours]])[0]

if res == 1:
   print(f"Based on your {hours} you're passed") 
else: 
   print(f"Based on your {hours} you're Failed")
```

```python
from sklearn.neighbors import KNeighborsClassifier

X = [
    [200 , 10],
    [280 , 12.5],
    [350 , 14.5],
    [400 , 15],
    [450 , 17],
    [500 , 18.5],
]
y = [0,0,0,1,1,1]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)

weight = float(input("Enter the weight : "))
size = float(input("Enter the Size : "))

prediction = model.predict([[weight,size]])[0]

if prediction == 0 :
    print("Based on the size and weight , Fruit can be Mango")
else:
    print("Based on the size and weight , Fruit can be Watermelon")
```
