# Practice with the Student performance dataset 

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import numpy as np

data = pd.read_csv('Student_Performance.csv')

X = data[['study_hours']]
y = data['overall_score']

model = LinearRegression()
model.fit(X,y)

predicted_score = model.predict(X)
mae = mean_absolute_error(y,predicted_score)
mse = mean_squared_error(y,predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y ,predicted_score)

print(f"Predicted Score is : {predicted_score}")
print("MAE Score is : " , round(mae,2))
print("MSE Score is : ",round(mse,2))
print("RMSE Score is : ",round(rmse,2))
print("r2 Score is : ",round(r2,4))
```
