# Basics of the Prompt Engineering 

Promts are contain 2 parts : 
    Parameteres ,
    Structure 


in Parameters :

1. Temperature : lower for creativity and deterministic result ; Higher for the diversity and the QA or Reasoning 
2. TOP P : Control model deteministic ; low for factual , high over diversity 
3. Max length : Manage you response length 

Components of the Good Prompts :

    1. Context:
    2. Instruction:
    3. Input Data:
    4. Output Indicator:

1. IN Context :
    additional context that can steet the model to better responses 
2. In Instruction :
    classify the feedback in category 
3. In Input Data :
    learn for the output from the input data by training 
4. In Output Data :
    what should be the output and in which shape
