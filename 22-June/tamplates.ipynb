# Tamplates 


1.  Explain Everythings

    Explain [TOPIC] to me as if I am a beginner.
    Start with simple concepts, then move to advanced ideas.
    Use examples, analogies, and practical applications.
    At the end, give me 3 key takeaways.


2. Deep research

Research [TOPIC] in depth.
Cover:

1. Background
2. Current trends
3. Advantages
4. Risks
5. Future outlook
   Use reliable sources where possible.


3. Problem Solving

Help me solve this problem:
[DESCRIBE PROBLEM]

Break it into smaller parts.
Identify root causes.
Suggest multiple solutions ranked by effectiveness.


4. Decision Making

    Help me decide between:
    Option A: [A]
    Option B: [B]

         Compare based on:

         - Cost
         - Time
         - Risk
         - Long-term value

         Recommend the best option.


5.  Planning

    Create a step-by-step plan to achieve: [GOAL]

        Constraints:
        [TIME / BUDGET / SKILLS]

        Include milestones and deadlines.


6. Learning Roadmap

Create a roadmap to learn [SKILL].
Divide into beginner, intermediate, advanced.
Suggest resources and practice exercises.

7. Prompt Improvement 

Improve this prompt to make AI responses more accurate:
[PROMPT]

Make it more specific and effective.

[AREA]

Challenge my thinking and suggest best practices.

[LIST]

[GOALS]

[TASK]
