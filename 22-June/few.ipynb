# Module 6 — Few-Shot Prompting

## Definition
Few-shot prompting means giving the model **multiple examples** before asking it to solve a task.

> Few-shot = multiple demonstrations + task

---

## When to Use
Best for:
- Complex tasks
- Domain-specific tasks
- Structured outputs
- Classification & scoring

---

## Example
```text
Review: Amazing
Label: Positive

Review: Terrible
Label: Negative

Review: Average
Label: Neutral

Review: Good but expensive
Label:
```

---

## Advantages
- Higher accuracy
- Better consistency
- Strong task adaptation

## Limitations
- Higher token cost
- Uses more context window

---

## Key Takeaway
Few-shot prompting improves model performance by teaching patterns through multiple examples.
