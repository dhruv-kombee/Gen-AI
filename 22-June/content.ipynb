# Module 7–10 Notes

## Module 7 — Instruction Prompting
Giving clear instructions to the LLM.

**Best Practices:**
- Use action verbs  
- Add constraints  
- Define output format  

**Key:** Clear instructions improve response quality.

---

## Module 8 — Role Prompting
Assigning a role/persona to the model.

Example:
```text
You are a senior AI engineer.
```

**Benefits:**
- Better expertise  
- Better reasoning  
- Controlled tone  

**Key:** Roles improve specialization.

---

## Module 9 — Context Engineering
Providing relevant information for better answers.

Examples:
- Documents  
- Transcripts  
- Search results  

**Key:** Better context reduces hallucinations.

---

## Module 10 — Output Formatting
Controlling response structure.

Formats:
- Markdown  
- JSON  
- Tables  

**Key:** Structured outputs improve consistency and automation.
