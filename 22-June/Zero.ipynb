# Module 4 — Zero-Shot Prompting

## Definition
Zero-shot prompting means asking an LLM to perform a task **without giving examples**.

> Zero-shot = Direct instruction only

The model uses knowledge learned during pretraining.

---

## When to Use
Best for simple and clear tasks:

- Summarization  
- Classification  
- Translation  
- Extraction  

Example:

```text
Summarize this article in 5 bullet points.
```

---

## Good Prompt Structure
Include:

- Task  
- Context  
- Constraints  
- Output Format  

Template:

```text
Task:
Context:
Constraints:
Output:
```

Example:

```text
Summarize this report in 5 bullet points focusing on risks.
```

---

## Best Practices
- Be specific  
- Add constraints  
- Define output format  
- Avoid ambiguity  

---

## Limitations
Zero-shot struggles with:

- Complex tasks  
- Domain-specific tasks  
- Strict scoring/rubrics  

---

## Key Takeaway
Zero-shot prompting is fast and efficient for clear tasks but becomes less reliable for complex problems.
