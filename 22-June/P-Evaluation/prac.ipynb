```python
import pandas as pd
import json
from statistics import mean
```

## Dataset Description

Dataset contains structured test cases for prompt evaluation.

Columns:
- id → Test case ID
- input → User query
- expected → Ground truth / expected behavior

```python
dataset = [
    {
        "id": 1,
        "input": "Summarize the meeting and extract action items.",
        "expected": "Summary with action items"
    },
    {
        "id": 2,
        "input": "Explain what transformer architecture is.",
        "expected": "Correct explanation"
    },
    {
        "id": 3,
        "input": "Convert meeting notes into JSON format.",
        "expected": "Valid JSON"
    }
]

df = pd.DataFrame(dataset)
df
```

```python
prompt_v1 = """
Answer the query.
"""

prompt_v2 = """
Answer clearly with structured formatting and important details.
"""

prompt_v3 = """
Answer clearly, include all required details, avoid hallucination, and follow output format strictly.
"""
```

## Response Collection

Run each prompt against every dataset input and store responses.

Example:
- Prompt V1 response
- Prompt V2 response
- Prompt V3 response

```python
responses = {
    "v1": [4, 3, 4],
    "v2": [4, 4, 5],
    "v3": [5, 5, 5]
}
```

# Scoring Criteria

Each response is evaluated using:

1. Accuracy (40%)
2. Relevance (25%)
3. Completeness (25%)
4. Format Compliance (10%)

Scoring Scale:
1 = Poor
2 = Weak
3 = Average
4 = Good
5 = Excellent

```python
def calculate_average(scores):
    return round(mean(scores), 2)

```

```python
benchmark = {}

for version, scores in responses.items():
    benchmark[version] = calculate_average(scores)

benchmark
```

```python
benchmark_df = pd.DataFrame(
    list(benchmark.items()),
    columns=["Prompt Version", "Average Score"]
)

benchmark_df
```

# Hallucination Measurement

Hallucination occurs when the LLM generates false or unsupported information.

Formula:

Hallucination Rate =
False Claims / Total Claims

```python
false_claims = 2
total_claims = 20

hallucination_rate = false_claims / total_claims
hallucination_rate
```

# Consistency Testing

Run the same prompt multiple times on the same input.

Check:
- Output variation
- Missing facts
- Formatting changes

```python
run_scores = [4.8, 4.6, 4.7, 4.9, 4.8]
consistency_score = round(mean(run_scores), 2)

consistency_score
```
