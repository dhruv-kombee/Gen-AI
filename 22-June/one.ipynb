# Module 5 — One-Shot Prompting

## Definition
One-shot prompting means giving the model **one example** before asking it to solve a task.

> One-shot = 1 example + new task

---

## When to Use
Best when:
- Zero-shot is inconsistent
- Format matters
- Task needs pattern guidance

---

## Example
```text
Input: Great service
Sentiment: Positive

Input: Bad support
Sentiment:
```

---

## Advantages
- Better consistency
- Improved accuracy
- Stronger formatting control

## Limitations
Struggles with complex or highly varied tasks.

---

## Key Takeaway
Use one-shot prompting when one example is enough to teach the output pattern.
